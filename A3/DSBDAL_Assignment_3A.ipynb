```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('BankChurners.csv')
```

###Initial Survey

```python
display(df.head())
df.describe()
```

```python
df.info()
df.isnull().sum()
```

```python
bins = [18, 30, 40, 50, 60, 100]
labels = ["18-30", "31-40", "41-50", "51-60", "60+"]

df["Age_Group"] = pd.cut(df["Customer_Age"], bins=bins, labels=labels)
```

```python
summary = df.groupby("Age_Group")["Credit_Limit"].agg(
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max",
    Standard_Deviation="std"
)

summary
```

```python
summary_statistics = df.groupby("Gender")["Credit_Limit"].agg(
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max",
    Standard_Deviation="std"
)

summary_statistics
```

```python
plt.figure()
df.boxplot(column="Credit_Limit", by="Age_Group")

plt.title("Credit Limit Distribution by Age Group")
plt.suptitle("")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

```python
plt.figure()
df.boxplot(column="Credit_Limit", by="Gender")

plt.title("Credit Limit Distribution by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```

```python
sns.violinplot(x="Age_Group", y="Credit_Limit", data=df)

plt.title("Credit Limit Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

```python
sns.violinplot(x="Gender", y="Credit_Limit", data=df)

plt.title("Credit Limit Distribution by Gender")
plt.show()
```

```python
plt.figure(figsize=(8,5))

for group in df["Age_Group"].unique():
    plt.hist(df[df["Age_Group"] == group]["Credit_Limit"],
             alpha=0.6,
             label=str(group))

plt.xlabel("Credit Limit")
plt.ylabel("Frequency")
plt.title("Credit Limit Distribution by Age Group")
plt.legend(title="Age Group")

plt.show()

```

```python
plt.hist(df[df["Gender"] == "F"]["Credit_Limit"], alpha=0.6, label="Female")
plt.hist(df[df["Gender"] == "M"]["Credit_Limit"], alpha=0.6, label="Male")

plt.xlabel("Credit Limit")
plt.ylabel("Frequency")
plt.title("Credit Limit Distribution by Gender")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,5))

for group in df["Age_Group"].unique():

    data = np.sort(df[df["Age_Group"] == group]["Credit_Limit"])
    y = np.arange(1, len(data) + 1) / len(data)

    plt.plot(data, y, label=str(group))

plt.xlabel("Credit Limit")
plt.ylabel("Cumulative Probability")
plt.title("ECDF of Credit Limit by Age Group")
plt.legend(title="Age Group")

plt.show()
```

```python
female = np.sort(df[df["Gender"] == "F"]["Credit_Limit"])
male = np.sort(df[df["Gender"] == "M"]["Credit_Limit"])

y_f = np.arange(1, len(female)+1) / len(female)
y_m = np.arange(1, len(male)+1) / len(male)

plt.figure()
plt.plot(female, y_f, label="Female")
plt.plot(male, y_m, label="Male")

plt.xlabel("Credit Limit")
plt.ylabel("Cumulative Probability")
plt.title("ECDF of Credit Limit by Gender")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.stripplot(x="Age_Group", y="Credit_Limit", data=df, jitter=True)

plt.title("Individual Credit Limits by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

```python
plt.figure()
sns.stripplot(x="Gender", y="Credit_Limit", data=df, jitter=True)

plt.title("Individual Credit Limits by Gender")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.stripplot(
    x="Age_Group",
    y="Credit_Limit",
    data=df,
    jitter=0.35,
    dodge=True,
    size=4,
    alpha=0.5
)

plt.title("Spread Strip Plot of Credit Limit by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.stripplot(
    x="Gender",
    y="Credit_Limit",
    data=df,
    jitter=0.35,
    dodge=True,
    size=4,
    alpha=0.5
)

plt.title("Spread Strip Plot of Credit Limit by Gender")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```
