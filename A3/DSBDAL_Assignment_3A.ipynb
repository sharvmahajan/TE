###Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('BankChurners.csv')
```

###Initial Survey

```python
display(df.head())
df.describe()
```

```python
df.info()
df.isnull().sum()
```

Here, we are dividing customers into different age groups instead of keeping exact ages.

The bins list sets the age ranges like 18–30, 31–40, and so on.

The labels give meaningful names to each of these ranges.

Using pd.cut(), each customer’s age is placed into the correct age group.

Finally, a new column called Age_Group is added to the dataframe, which helps in better analysis based on age categories.

```python
bins = [18, 30, 40, 50, 60, 100]
labels = ["18-30", "31-40", "41-50", "51-60", "60+"]

df["Age_Group"] = pd.cut(df["Customer_Age"], bins=bins, labels=labels)
```

The data is first grouped based on the Age_Group column.

For each age category, useful statistics of the Credit_Limit such as the average, median, smallest, largest, and standard deviation are shown.

```python
summary = df.groupby("Age_Group")["Credit_Limit"].agg(
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max",
    Standard_Deviation="std"
)

summary
```

The dataset is grouped using the Gender column.

For each gender category, key statistics of the Credit_Limit, including the mean, median, minimum, maximum, and standard deviation are shown.

```python
summary_statistics = df.groupby("Gender")["Credit_Limit"].agg(
    Mean="mean",
    Median="median",
    Minimum="min",
    Maximum="max",
    Standard_Deviation="std"
)

summary_statistics
```

Boxplot shown for distribution by Age Group



```python
plt.figure()
df.boxplot(column="Credit_Limit", by="Age_Group")

plt.title("Credit Limit Distribution by Age Group")
plt.suptitle("")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

Boxplot shown for distribution by Gender

```python
plt.figure()
df.boxplot(column="Credit_Limit", by="Gender")

plt.title("Credit Limit Distribution by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```

Violin Plot for Distribution by Age Group

```python
sns.violinplot(x="Age_Group", y="Credit_Limit", data=df)

plt.title("Credit Limit Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

Violin Plot for distribution by Gender

```python
sns.violinplot(x="Gender", y="Credit_Limit", data=df)

plt.title("Credit Limit Distribution by Gender")
plt.show()
```

Histogram for distribution by Age Group

```python
plt.figure(figsize=(8,5))

for group in df["Age_Group"].unique():
    plt.hist(df[df["Age_Group"] == group]["Credit_Limit"],
             alpha=0.6,
             label=str(group))

plt.xlabel("Credit Limit")
plt.ylabel("Frequency")
plt.title("Credit Limit Distribution by Age Group")
plt.legend(title="Age Group")

plt.show()

```

Histogram fro distribution by Gender

```python
plt.hist(df[df["Gender"] == "F"]["Credit_Limit"], alpha=0.6, label="Female")
plt.hist(df[df["Gender"] == "M"]["Credit_Limit"], alpha=0.6, label="Male")

plt.xlabel("Credit Limit")
plt.ylabel("Frequency")
plt.title("Credit Limit Distribution by Gender")
plt.legend()

plt.show()
```

ECDF (Empirical Cumulative Distribution Function) shows cumulative distribution across age groups



```python
plt.figure(figsize=(8,5))

for group in df["Age_Group"].unique():

    data = np.sort(df[df["Age_Group"] == group]["Credit_Limit"])
    y = np.arange(1, len(data) + 1) / len(data)

    plt.plot(data, y, label=str(group))

plt.xlabel("Credit Limit")
plt.ylabel("Cumulative Probability")
plt.title("ECDF of Credit Limit by Age Group")
plt.legend(title="Age Group")

plt.show()
```

ECDF (Empirical Cumulative Distribution Function) shows cumulative distribution across gender



```python
female = np.sort(df[df["Gender"] == "F"]["Credit_Limit"])
male = np.sort(df[df["Gender"] == "M"]["Credit_Limit"])

y_f = np.arange(1, len(female)+1) / len(female)
y_m = np.arange(1, len(male)+1) / len(male)

plt.figure()
plt.plot(female, y_f, label="Female")
plt.plot(male, y_m, label="Male")

plt.xlabel("Credit Limit")
plt.ylabel("Cumulative Probability")
plt.title("ECDF of Credit Limit by Gender")
plt.legend()

plt.show()
```

Strip Plot for distribution across Age Group

```python
plt.figure(figsize=(8,5))

sns.stripplot(x="Age_Group", y="Credit_Limit", data=df, jitter=True)

plt.title("Individual Credit Limits by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```

Strip Plot for distribution across Gender

```python
plt.figure()
sns.stripplot(x="Gender", y="Credit_Limit", data=df, jitter=True)

plt.title("Individual Credit Limits by Gender")
plt.xlabel("Gender")
plt.ylabel("Credit Limit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.stripplot(
    x="Age_Group",
    y="Credit_Limit",
    data=df,
    jitter=0.35,
    dodge=True,
    size=4,
    alpha=0.5
)

plt.title("Spread Strip Plot of Credit Limit by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Credit Limit")

plt.show()
```
