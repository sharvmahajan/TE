### Importing libraries

```python
import pandas as pd
```

### Load Dataset

```python
df = pd.read_csv("covid_vaccine_statewise.csv")
print("First 5 rows:")
df.head()
```

### Basic dataset information

```python
# a) Describe dataset
print("Dataset Info:")
df.info()
```

```python
df.isnull().sum()
```

```python
print("Statistical Summary:")
df.describe()
```

### Data Preprocessing

```python
# Clean column names
df.columns = df.columns.str.strip()
```

```python
# Convert numeric columns
numeric_cols = [
    'First Dose Administered',
    'Second Dose Administered',
    'Male (Doses Administered)',
    'Female (Doses Administered)'
]

for col in numeric_cols:
  df[col] = pd.to_numeric(df[col], errors='coerce')


print("Numeric Columns after conversion:")
display(df[numeric_cols].head())
```

```python
# Remove rows where State = 'India'
india_rows_count = df[df['State'] == 'India'].shape[0]
print(f"Number of rows where State is 'India': {india_rows_count}")
state_df = df[df['State'] != 'India']
print(f"Number of rows after removing 'India': {state_df.shape[0]}")
```

### Data Extraction

```python
# b) First dose state-wise
print("First Dose (State-wise):")
first_dose = state_df.groupby('State')['First Dose Administered'].max().astype(int)
first_dose
```

```python
# c) Second dose state-wise
print("\nSecond Dose (State-wise):")
second_dose = state_df.groupby('State')['Second Dose Administered'].max().astype(int)
second_dose
```

```python
# d) Total males vaccinated
total_males = int(df['Male (Doses Administered)'].max())
print("Total Males Vaccinated:", total_males)
```

```python
# e) Total females vaccinated
total_females = int(df['Female (Doses Administered)'].max())
print("Total Females Vaccinated:", total_females)
```

### Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Top 10 states for First Dose Administered
top_10_first_dose = first_dose.nlargest(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_first_dose.values, y=top_10_first_dose.index, palette='viridis', hue=top_10_first_dose.index, legend=False)
plt.title('Top 10 States by First Dose Administered')
plt.xlabel('Number of First Doses Administered')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

```python
print(total_males+total_females)
```

```python
# Top 10 states for Second Dose Administered
top_10_second_dose = second_dose.nlargest(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_second_dose.values, y=top_10_second_dose.index, palette='magma', hue=top_10_second_dose.index, legend=False)
plt.title('Top 10 States by Second Dose Administered')
plt.xlabel('Number of Second Doses Administered')
plt.ylabel('State')
plt.tight_layout()
plt.show()
```

```python
# Total Male vs Female Vaccinations
vaccination_gender = pd.DataFrame({
    'Gender': ['Male', 'Female'],
    'Total Doses Administered': [total_males, total_females]
})

plt.figure(figsize=(8, 5))
sns.barplot(x='Gender', y='Total Doses Administered', data=vaccination_gender, hue='Gender', legend=False)
plt.title('Total Male vs Female Vaccinations')
plt.ylabel('Total Doses Administered')
plt.tight_layout()
plt.show()
```

```python
# Calculate total male and female doses administered per state
state_gender_vaccination = state_df.groupby('State')[['Male (Doses Administered)', 'Female (Doses Administered)']].max().reset_index()

# Calculate total doses for ranking
state_gender_vaccination['Total Doses'] = state_gender_vaccination['Male (Doses Administered)'] + state_gender_vaccination['Female (Doses Administered)']

# Sort by total doses to get top states
top_states_gender_vaccination = state_gender_vaccination.nlargest(10, 'Total Doses')

# Melt the DataFrame for easier plotting with seaborn
top_states_melted = top_states_gender_vaccination.melt(id_vars='State', value_vars=['Male (Doses Administered)', 'Female (Doses Administered)'], var_name='Gender', value_name='Total Doses Administered')

# Rename the 'Gender' column values for cleaner legend labels
top_states_melted['Gender'] = top_states_melted['Gender'].replace({
    'Male (Doses Administered)': 'Male',
    'Female (Doses Administered)': 'Female'
})

plt.figure(figsize=(14, 7))
sns.barplot(x='State', y='Total Doses Administered', hue='Gender', data=top_states_melted, palette={'Male': 'blue', 'Female': 'brown'})
plt.title('Top 10 States: Male vs Female Total Vaccinations')
plt.xlabel('State')
plt.ylabel('Total Doses Administered')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Gender') # Legend labels will now be automatically picked from the 'Gender' column
plt.tight_layout()
plt.show()
```
